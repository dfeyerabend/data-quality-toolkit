# Pandas engine: messy CSV check

Run the pandas check on a small, messy CSV and inspect the overview table and the full per column report. The pandas engine infers column types, so it also flags wrong types that a CSV hides.

```python
import os, sys
# Make the project root importable (engines, utils) no matter where the kernel starts.
_root = os.getcwd()
while not os.path.isdir(os.path.join(_root, "engines")) and _root != os.path.dirname(_root):
    _root = os.path.dirname(_root)
sys.path.insert(0, _root)

import warnings
warnings.filterwarnings("ignore")

import pandas as pd

from engines.core import inspect_data
from utils.notebook_utils import show_struct

DATA_PATH = os.path.join(_root, "example_data", "example-sales_data.csv")
df = pd.read_csv(DATA_PATH)
```

## The dataset
A small, messy sales CSV: wrong types, dirty values and inconsistent formats.

```python
print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
print(f"Column names: {df.columns.tolist()}")
```

## Run the check
One call runs the check and returns an overview table and a detailed report.

```python
summary_table, report = inspect_data(df, engine="pandas")
```

## Overview table
One row per column: missing values, outliers and uniqueness at a glance.

```python
summary_table
```

## Full report per column
Expand any column to see its detailed diagnostics.

```python
show_struct(report)
```
